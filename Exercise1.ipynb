# Problem 1
Write code to print the first 10 natural numbers to the console using a while loop.

```python
def print_n_to_m(n, m):
    while n <= m:
        print(n, end=" ")
        n = n + 1
        
n = 2
m = 10

print_n_to_m(n, m)

print("\n")

print_n_to_m(1,5)
```

# Problem 5
Write code to display only specific items from a list/array that satisfy the following conditions:

• The number must be divisible by five.

• If the number is greater than or equal to 150, then skip it.

• If the number is greater than 500, then stop the loop.

Example input:
• Python: [12, 75, 150, 182, 145, 525, 50]

```python
# Creating a list
list1 = [12, 75, 150, 182, 145, 525, 50]

for i in list1:
    if i > 500:
        break
    elif i >= 150:
        continue
    elif i % 5 == 0:
        print(i)

for i in list1:
    if i % 5 == 0 and i < 150:
        print(i)
    elif i > 500:
        break
```

# Problem 6
Write code to count the digits of a given number n. The code should print the result to the console.
For example, for n = 75869 the output should be 5.

```python

n = 75869

def count_digits(n):
    count = 0
    while n > 0:
        n = n // 10
        count = count + 1
    return count

print(count_digits(n))

n_str = str(n)
print(len(n_str))
```

# Problem 7
Write code to print all prime numbers between any given pair of numbers n and m to the console.
Use as example input n = 25 and m = 50.

NOTE:A prime number is a natural number greater than 1 that can only be divided by itself and
1 without remainders. Examples are: 2, 3, 5, 7, 11.

```python
n = 25
m = 50

while n <= m:
    is_prime = True
    divisor = 2
    
    while divisor < n:
        if n % divisor == 0:
            is_prime = False
        divisor = divisor + 1
    
    if is_prime == True:
        print(n)
   
    n = n + 1

```

```python
# start = 0 inclusive
# stop = 10  exclusive
# step = 1     
n = 25
m = 50
for i in range(n, m + 1):
    is_prime = True
    
    for divisor in range(2, i):
        if i % divisor == 0:
            is_prime = False
            break
    
    if is_prime == True:
        print(i)
```
